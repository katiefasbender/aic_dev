Demos for hdps.catalogs, which will contain functionality to build the various parts of the **Argus Input Catalog (AIC)**:

**AIC Refcat** (a reference catalog of known sources, built by AICRefcat)
- "constant" "static" point sources & extended objects to save in Argus observations
- will be constructed from multiple constituent reference catalogs (further testing required as of 2026/05):
    - ATLAS-Refcat2 Stellar Reference Catalog
        - for sources g <= 19 mag
        - only contains stars (with some galaxy contamination)
    - Gaia Galaxy Candidates
        - for sources G < 19 mag (still needs testing)
        - to supplement ATLAS-Refcat2, which does not contain galaxies
    - PanSTARRS
        - for sources g > 19 mag
        - contains both stars and galaxies
- Built by the `AICRefcatBuilder` class, loaded by `AICRefcat` class.  `AICRefcatBuilder` does the following:
    - loads all sources of interest with lsdb
    - identifies minipix/tile IDs for each source, if not already calculated & saved
    - writes catalog with IDs to disk, if not already saved


*in development:*

**AIC Base** (a minipix catalog, will be built by `AICBase`)
- "always-on" minipix containing "constant" "static" point sources & extended objects from AIC Refcat
- `AICBase` builds this; This class does the following:
    - reads in AIC Refcat minipix/tile IDS
    - finds unique indices; save the list?
    - option to activate those minipix 

**AIC SSOs** (a minipix catalog)
- "toggle" minipix containing SSO detections predicted by pipeline nightly 

**AIC Transients** (a minipix catalog)
- "toggle" minipix containing transient events flagged by the pipeline, can be added to AIC Base
